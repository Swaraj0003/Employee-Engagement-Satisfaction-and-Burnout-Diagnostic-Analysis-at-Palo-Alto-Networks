```python

# STEP 9: Satisfaction Analysis

satisfaction_columns = [
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "RelationshipSatisfaction",
    "JobInvolvement",
    "WorkLifeBalance"
]

available_columns = [
    column for column in satisfaction_columns
    if column in df.columns
]

for column in available_columns:

    print(f"\n{column} rating percentages:")

    percentages = (
        df[column]
        .value_counts(normalize=True)
        .sort_index()
        .mul(100)
        .round(2)
    )

    print(percentages)

    plt.figure(figsize=(7, 4))

    sns.countplot(
        data=df,
        x=column,
        order=[1, 2, 3, 4]
    )

    plt.title(f"{column} Rating Distribution")
    plt.xlabel("Rating (1 = Low, 4 = High)")
    plt.ylabel("Employee Count")

    plt.tight_layout()

    plt.savefig(f"plots/{column}_ratings.png")

    plt.show()
    plt.close()

print("Satisfaction analysis completed!")
```

```python

import json
import numpy as np
import pandas as pd
from pathlib import Path

# Load the original dataset
df = pd.read_csv("employee_data.csv")
df.columns = df.columns.str.strip()

# Convert binary/category fields into consistent labels
if "Attrition" in df.columns:
    df["Attrition"] = (
        df["Attrition"].astype("string").str.strip().str.lower()
        .map({
            "yes": "Yes", "no": "No",
            "1": "Yes", "0": "No",
            "true": "Yes", "false": "No"
        })
        .fillna(df["Attrition"])
    )

if "OverTime" in df.columns:
    df["OverTime"] = (
        df["OverTime"].astype("string").str.strip().str.lower()
        .map({"yes": "Yes", "no": "No"})
        .fillna(df["OverTime"])
    )

# Build the engagement index
engagement_cols = [
    "JobInvolvement",
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "RelationshipSatisfaction"
]

if all(c in df.columns for c in engagement_cols):
    ratings = df[engagement_cols].apply(
        pd.to_numeric, errors="coerce"
    )
    valid = ratings.notna().all(axis=1)
    mean_rating = ratings.mean(axis=1)

    df["Engagement_Index"] = np.nan
    df.loc[valid, "Engagement_Index"] = (
        (mean_rating[valid] - 1) / 3 * 100
    )

# Work-life balance index
if "WorkLifeBalance" in df.columns:
    df["WorkLifeBalance"] = pd.to_numeric(
        df["WorkLifeBalance"], errors="coerce"
    )
    df["WLB_Index"] = (
        (df["WorkLifeBalance"] - 1) / 3 * 100
    )

# Transparent workload warning indicators
if {"OverTime", "WorkLifeBalance"}.issubset(df.columns):
    df["Overtime_Flag"] = (
        df["OverTime"].astype("string").str.lower().eq("yes")
    )
    df["Low_WLB_Flag"] = df["WorkLifeBalance"].le(2)

    df["Combined_Warning_Flag"] = (
        df["Overtime_Flag"] & df["Low_WLB_Flag"]
    )

# Calculate the project KPIs
kpis = {
    "employee_count": int(len(df)),
    "column_count": int(len(df.columns)),
    "missing_cells": int(df.isna().sum().sum()),
    "duplicate_rows": int(df.duplicated().sum())
}

if "Engagement_Index" in df.columns:
    kpis["mean_engagement_index"] = round(
        float(df["Engagement_Index"].mean()), 2
    )

if "WLB_Index" in df.columns:
    kpis["mean_work_life_balance_index"] = round(
        float(df["WLB_Index"].mean()), 2
    )

if "Overtime_Flag" in df.columns:
    kpis["overtime_percentage"] = round(
        float(df["Overtime_Flag"].mean() * 100), 2
    )

if "Combined_Warning_Flag" in df.columns:
    kpis["combined_warning_percentage"] = round(
        float(df["Combined_Warning_Flag"].mean() * 100), 2
    )

if "Attrition" in df.columns:
    attrition_known = df["Attrition"].isin(["Yes", "No"])
    kpis["attrition_rate_percentage"] = round(
        float(df.loc[attrition_known, "Attrition"].eq("Yes").mean() * 100),
        2
    ) if attrition_known.any() else None

# Department-level results
if "Department" in df.columns:
    agg = {}
    if "Engagement_Index" in df.columns:
        agg["Engagement_Index"] = "mean"
    if "WorkLifeBalance" in df.columns:
        agg["WorkLifeBalance"] = "mean"
    if "Combined_Warning_Flag" in df.columns:
        agg["Combined_Warning_Flag"] = lambda s: s.mean() * 100

    if agg:
        department_summary = df.groupby("Department").agg(agg).round(2)
        display(department_summary)
        department_summary.to_csv("department_kpis.csv")

# Save reusable outputs
Path("project_outputs").mkdir(exist_ok=True)

df.to_csv("project_outputs/analytical_dataset.csv", index=False)

with open("project_outputs/kpis.json", "w", encoding="utf-8") as f:
    json.dump(kpis, f, indent=2, allow_nan=False)

print("Project KPI calculation completed.")
print(json.dumps(kpis, indent=2))
```
